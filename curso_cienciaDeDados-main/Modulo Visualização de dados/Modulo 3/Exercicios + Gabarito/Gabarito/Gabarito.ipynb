## Exercício — Seaborn e Análise Estatística

Neste exercício, usamos uma base simples chamada `alunos` para praticar os principais gráficos do Seaborn:

- distribuição com histograma e KDE;
- comparação entre grupos com boxplot;
- visualização dos pontos com stripplot;
- relação entre variáveis com scatter plot;
- correlação com heatmap.

O foco é entender a função de cada gráfico e como o parâmetro `hue` ajuda a separar visualmente os grupos.

```
# ✅ Célula de setup — execute esta primeiro!
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Deixando os resultados sempre iguais
np.random.seed(42)

# Criando uma base simples de alunos
n = 60
alunos = pd.DataFrame({
    'turma': np.random.choice(['A', 'B'], size=n),
    'horas_estudo': np.random.normal(5, 1.5, size=n).clip(1, 9),
    'faltas': np.random.poisson(3, size=n),
})

# Criando uma nota com relação simples com horas de estudo e faltas
alunos['nota'] = (
    5 + alunos['horas_estudo'] * 0.7 - alunos['faltas'] * 0.25 + np.random.normal(0, 0.8, size=n)
).clip(0, 10)

alunos.head()
```

## 1. Histograma com KDE

O histograma mostra como as notas estão distribuídas.  
O `kde=True` adiciona uma curva suavizada, facilitando a leitura do formato da distribuição.

Usamos `hue='turma'` para comparar as turmas A e B no mesmo gráfico.

```
plt.figure(figsize=(8, 4))
sns.histplot(data=alunos, x='nota', hue='turma', kde=True)
plt.title('Distribuição das notas por turma')
plt.xlabel('Nota')
plt.ylabel('Quantidade de alunos')
plt.show()
```

## 2. Boxplot com Stripplot

O boxplot ajuda a enxergar a mediana, a variação dos dados e possíveis outliers.

O stripplot mostra cada aluno como um ponto. Assim, o gráfico fica mais fácil de interpretar, porque vemos tanto o resumo estatístico quanto os dados reais.

```
plt.figure(figsize=(7, 4))
sns.boxplot(data=alunos, x='turma', y='nota')
sns.stripplot(data=alunos, x='turma', y='nota', jitter=True)
plt.title('Notas por turma')
plt.xlabel('Turma')
plt.ylabel('Nota')
plt.show()
```

## 3. Scatter plot com hue

O scatter plot mostra a relação entre duas variáveis numéricas.

Aqui, usamos `horas_estudo` no eixo X e `nota` no eixo Y. A separação por `hue='turma'` ajuda a comparar se as turmas têm comportamento parecido.

```
plt.figure(figsize=(8, 4))
sns.scatterplot(data=alunos, x='horas_estudo', y='nota', hue='turma')
plt.title('Relação entre horas de estudo e nota')
plt.xlabel('Horas de estudo')
plt.ylabel('Nota')
plt.show()
```

## 4. Heatmap de correlação

A matriz de correlação mostra a força da relação entre variáveis numéricas.

Valores próximos de **1** indicam relação positiva forte.  
Valores próximos de **-1** indicam relação negativa forte.  
Valores próximos de **0** indicam pouca relação linear.

```
correlacao = alunos[['horas_estudo', 'faltas', 'nota']].corr()

plt.figure(figsize=(6, 4))
sns.heatmap(correlacao, annot=True, cmap='Blues', vmin=-1, vmax=1)
plt.title('Matriz de correlação')
plt.show()
```

## Conclusão

Com esses gráficos, conseguimos fazer uma análise simples e completa:

- O histograma mostra a distribuição das notas;
- O boxplot compara as turmas;
- O stripplot mostra os alunos individualmente;
- O scatter plot mostra a relação entre estudo e nota;
- O heatmap resume as correlações entre as variáveis numéricas.
