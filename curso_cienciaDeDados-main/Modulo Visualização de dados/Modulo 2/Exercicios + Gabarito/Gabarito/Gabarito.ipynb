## Exercício — Layout com GridSpec

**Desafio:** montar um layout com `GridSpec` contendo:

- **Coluna esquerda:** um gráfico de linha ocupando as **2 linhas**;
- **Coluna direita, linha 0:** um **scatter plot**;
- **Coluna direita, linha 1:** um **histograma**.

O foco do exercício é entender a **estrutura do layout**, não os dados utilizados nos gráficos. Por isso, podemos gerar dados aleatórios com `np.random`.

## Resposta explicada

Para resolver o problema, usamos o `GridSpec` para dividir a figura em uma grade de **2 linhas por 2 colunas**.

A lógica principal é:

1. Criar uma figura com `plt.figure()`;
2. Criar uma grade com `gridspec.GridSpec(2, 2)`;
3. Fazer o gráfico da esquerda ocupar as duas linhas usando `gs[:, 0]`;
4. Fazer o gráfico superior direito ocupar a posição `gs[0, 1]`;
5. Fazer o gráfico inferior direito ocupar a posição `gs[1, 1]`.

A parte mais importante é entender a indexação do `GridSpec`:

- `gs[:, 0]` → todas as linhas da coluna 0;
- `gs[0, 1]` → linha 0, coluna 1;
- `gs[1, 1]` → linha 1, coluna 1.

````python
# Importação das bibliotecas necessárias
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.gridspec as gridspec
````

````python
# Criando a figura principal
fig = plt.figure(figsize=(12, 5), constrained_layout=True)

# Criando uma grade com 2 linhas e 2 colunas
gs = gridspec.GridSpec(2, 2, figure=fig)

# Gráfico 1: linha ocupando as duas linhas da coluna esquerda
ax_1 = fig.add_subplot(gs[:, 0])
ax_1.plot(np.random.rand(50).cumsum(), color='blue')
ax_1.set_title('Gráfico de linha - ocupa 2 linhas')
ax_1.set_xlabel('Eixo X')
ax_1.set_ylabel('Valores acumulados')

# Gráfico 2: scatter plot na parte superior direita
ax_2 = fig.add_subplot(gs[0, 1])
ax_2.scatter(np.random.rand(20), np.random.rand(20), color='orange')
ax_2.set_title('Scatter plot')
ax_2.set_xlabel('X')
ax_2.set_ylabel('Y')

# Gráfico 3: histograma na parte inferior direita
ax_3 = fig.add_subplot(gs[1, 1])
ax_3.hist(np.random.randn(100), bins=15, color='green')
ax_3.set_title('Histograma')
ax_3.set_xlabel('Valores')
ax_3.set_ylabel('Frequência')

plt.show()
````

## Por que essa solução está correta?

O exercício pede um layout específico. A solução atende ao pedido porque:

- O gráfico da esquerda usa `gs[:, 0]`, então ele ocupa **todas as linhas** da primeira coluna;
- O scatter plot usa `gs[0, 1]`, ficando na **linha superior da coluna direita**;
- O histograma usa `gs[1, 1]`, ficando na **linha inferior da coluna direita**;
- Os dados foram gerados com `np.random`, como solicitado no enunciado.

Portanto, o mais importante do exercício não é o tipo exato dos dados, mas sim a organização correta dos gráficos dentro da grade.
